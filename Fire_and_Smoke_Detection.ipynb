```python
import gradio as gr
from ultralytics import YOLO
import cv2
import numpy as np
from PIL import Image

model = YOLO('/content/runs/detect/train/weights/best.pt')

def predict_image(img):
    img_np = np.array(img)
    results = model.predict(source=img_np, save=False, conf=0.25)

    for r in results:
        im_array = r.plot()
        output_img = Image.fromarray(im_array[..., ::-1])

    return output_img

iface = gr.Interface(
    fn=predict_image,
    inputs=gr.Image(type="pil", label="Upload Image"),
    outputs=gr.Image(type="pil", label="Detected Objects"),
    title="YOLO Detection",
    description="Upload an image for detection.",
    live=False
)

# FIX event-loop bug
iface.queue(False)

iface.launch(debug=False)
```
